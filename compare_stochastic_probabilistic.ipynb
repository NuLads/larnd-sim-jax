```python
# Initialize Strategies
stoch_strategy = LUTSimulation(response)
prob_strategy = LUTProbabilisticSimulation(response)
```

```python
# Run Probabilistic Simulation
print("Running probabilistic simulation...")
prob_prediction = prob_strategy.predict(current_params, tracks, track_fields, 42)

unique_pixels = prob_prediction['unique_pixels']

all_tick_probs = {}
all_adcs_expected = {}

for pix_idx, target_pixel in enumerate(unique_pixels):
    if target_pixel < 0:
        continue
    all_tick_probs[int(target_pixel)] = jnp.exp(prob_prediction['hit_prob'][pix_idx, 0, :])
    all_adcs_expected[int(target_pixel)] = prob_prediction['adcs_distrib'][pix_idx, 0, :]

print(f"Stored probabilistic expected distributions for {len(all_tick_probs)} valid pixels.")
```

```python
# Run Stochastic Simulation Multiple Times (Optimized with Chunked JAX vmap)
from larndsim.fee_jax import get_adc_values, digitize

N_RUNS = 10000
CHUNK_SIZE = 200  # Process in chunks to prevent memory blow-up
print(f"Running stochastic simulation {N_RUNS} times using chunked JAX vmap...")

# Extract the waveforms that were already simulated
wfs = prob_prediction['wfs']
unique_pixels = prob_prediction['unique_pixels']

# Generate random keys for each run
keys = jax.random.split(jax.random.PRNGKey(42), N_RUNS)

@jax.jit
def run_chunk_stochastic(keys_chunk):
    # Vectorize the stochastic ADC generation over a chunk of random seeds
    vmap_get_adc = jax.vmap(get_adc_values, in_axes=(None, None, 0))
    integrals, ticks = vmap_get_adc(current_params, wfs, keys_chunk)
    
    # Digitize is element-wise, so we just pass the whole array
    adcs = digitize(current_params, integrals)
    
    # We only care about the FIRST hit for each pixel (index 0 along the hits axis)
    first_ticks = ticks[:, :, 0] # Shape: (CHUNK_SIZE, Npix)
    first_adcs = adcs[:, :, 0]   # Shape: (CHUNK_SIZE, Npix)
    
    return first_ticks, first_adcs

all_first_ticks = []
all_first_adcs = []

# Process in chunks to save memory but still drastically reduce Python overhead
for i in tqdm(range(0, N_RUNS, CHUNK_SIZE)):
    chunk_keys = keys[i:i+CHUNK_SIZE]
    ft, fa = run_chunk_stochastic(chunk_keys)
    all_first_ticks.append(ft)
    all_first_adcs.append(fa)

first_ticks_all = jnp.concatenate(all_first_ticks, axis=0)
first_adcs_all = jnp.concatenate(all_first_adcs, axis=0)

# Format outputs into dictionaries per pixel
all_stoch_ticks = {}
all_stoch_adcs = {}

for pix_idx, p in enumerate(unique_pixels):
    p_int = int(p)
    if p_int < 0:
        continue
        
    # A valid hit must have a tick within the waveform bounds
    valid_mask = first_ticks_all[:, pix_idx] < (wfs.shape[1] - 3)
    
    all_stoch_ticks[p_int]= np.array(first_ticks_all[valid_mask, pix_idx])
    all_stoch_adcs[p_int] = np.array(first_adcs_all[valid_mask, pix_idx])

print(f"Collected stochastic samples efficiently for {len(all_stoch_ticks)} pixels.")
```

## Tuning the Probabilistic Simulation Parameters

To close the gap between the stochastic and probabilistic simulations, we sweep the new parameters:
1. `excess_offset`: empirical offset in electrons to represent average excess charge at discretization trigger time and reset noise selection bias.
2. `smooth_sigma_scale`: factor scaling the effective uncorrelated noise standard deviation.

We evaluate each configuration by its Mean Absolute Error (MAE) of hit rate bias across all common pixels.

```python
# --- Cell: Tuning of excess_offset and smooth_sigma_scale ---
from larndsim.sim_jax import simulate_probabilistic
import numpy as np

# Define a grid to sweep
offsets = np.linspace(0.0, 400.0, 9)       # 0, 50, 100, 150, 200, 250, 300, 350, 400
scales = np.linspace(1.0, 1.4, 5)          # 1.0, 1.1, 1.2, 1.3, 1.4

# Get the common pixels and empirical rate
common_pixels_tune = sorted(set(all_tick_probs.keys()) & set(all_stoch_ticks.keys()))
emp_rate_tune = np.array([len(all_stoch_ticks[p]) / N_RUNS for p in common_pixels_tune])

best_mae = float('inf')
best_offset = 0.0
best_scale = 1.0
best_bias = 0.0
best_std = 0.0

print(f"{'Offset (e-)':12s} | {'Scale':6s} | {'Mean Bias':11s} | {'MAE Bias':10s} | {'Std Bias':10s}")
print("-" * 60)

for scale in scales:
    for offset in offsets:
        params_sweep = current_params.replace(excess_offset=offset, smooth_sigma_scale=scale)
        # Run only the probabilistic part of the simulation
        adcs_dist_sweep, _, _, ticks_prob_sweep, _ = simulate_probabilistic(params_sweep, wfs, unique_pixels)
        
        all_tick_probs_sweep = {}
        for pix_idx, target_pixel in enumerate(unique_pixels):
            if target_pixel < 0:
                continue
            all_tick_probs_sweep[int(target_pixel)] = np.array(jnp.exp(ticks_prob_sweep[pix_idx, 0, :]))
            
        lambda_prob_sweep = np.array([float(np.sum(all_tick_probs_sweep[p])) for p in common_pixels_tune])
        bias_sweep = lambda_prob_sweep - emp_rate_tune
        mean_bias = float(np.mean(bias_sweep))
        std_bias = float(np.std(bias_sweep))
        mae_bias = float(np.mean(np.abs(bias_sweep)))
        
        print(f"{offset:12.1f} | {scale:6.2f} | {mean_bias:+.6f} | {mae_bias:.6f} | {std_bias:.6f}")
        
        if mae_bias < best_mae:
            best_mae = mae_bias
            best_offset = offset
            best_scale = scale
            best_bias = mean_bias
            best_std = std_bias

print("-" * 60)
print(f"Best parameters found:")
print(f"  excess_offset      = {best_offset:.1f} e-")
print(f"  smooth_sigma_scale = {best_scale:.2f}")
print(f"  Mean Bias          = {best_bias:+.6f}")
print(f"  MAE Bias           = {best_mae:.6f}")

# Re-run probabilistic prediction with the best parameters to update downstream variables
print("\nRe-running probabilistic simulation with best parameters...")
current_params = current_params.replace(excess_offset=best_offset, smooth_sigma_scale=best_scale)
prob_prediction = prob_strategy.predict(current_params, tracks, track_fields, 42)

# Update the global tick probs and adcs expected dictionaries for downstream plotting
for pix_idx, target_pixel in enumerate(unique_pixels):
    if target_pixel < 0:
        continue
    all_tick_probs[int(target_pixel)] = jnp.exp(prob_prediction['hit_prob'][pix_idx, 0, :])
    all_adcs_expected[int(target_pixel)] = prob_prediction['adcs_distrib'][pix_idx, 0, :]
```

## Debugging: ADC / Charge Bias
For matched hits (same tick window), compare the expected ADC from the probabilistic model against the stochastic ADC distribution.

```python
# --- Cell D: ADC/charge bias (matched hits within ±TICK_TOL ticks) ---
# Fully vectorised: all operations are batched over (N_RUNS, N_pix) simultaneously.
# Uses first_ticks_all / first_adcs_all (N_RUNS, N_pix) from cell 5 directly.
TICK_TOL = 2   # half-window in ticks

# ── 1. Build dense pixel-indexed matrices (N_pix_common, N_ticks) ──────────────
# Map common_pixels → column index in first_ticks_all
pix_arr    = np.array(list(prob_prediction['unique_pixels']))  # ordered as in simulation
Nticks     = wfs.shape[1]

# t_prob_mat[i, t]  = P(tick=t | pixel i, hit-slot 0)     (N_pix, Nticks)
# a_exp_mat[i, t]   = expected ADC at tick t for pixel i   (N_pix, Nticks)
valid_pix_mask = pix_arr >= 0
N_pix_sim = int(valid_pix_mask.sum())

t_prob_mat = np.array(jnp.exp(prob_prediction['hit_prob'][:, 0, :]))  # (N_pix_sim_padded, Nticks)
a_exp_mat  = np.array(prob_prediction['adcs_distrib'][:, 0, :])

# Restrict to valid (non-padding) pixels
valid_indices = np.where(valid_pix_mask)[0]                # indices into simulation arrays
t_prob_mat = t_prob_mat[valid_indices]                     # (N_pix_sim, Nticks)
a_exp_mat  = a_exp_mat[valid_indices]                      # (N_pix_sim, Nticks)
valid_pix_ids = pix_arr[valid_indices]                     # pixel IDs in order

# Map common_pixels → their index in valid_pix_ids
common_pix_arr = np.array(common_pixels)
pix_to_col = {int(p): col for col, p in enumerate(valid_pix_ids)}
col_indices = np.array([pix_to_col[p] for p in common_pixels])  # (N_common,)

# ── 2. Slice first_ticks_all / first_adcs_all to common pixels ─────────────────
# first_ticks_all: (N_RUNS, N_pix_sim_padded) — columns ordered as unique_pixels
# We need to map col_indices back to the padded array columns
padded_pix_ids = list(prob_prediction['unique_pixels'])
padded_col = np.array([padded_pix_ids.index(p) for p in common_pixels])  # (N_common,)

stoch_ticks_mat = np.array(first_ticks_all[:, padded_col])   # (N_RUNS, N_common)
stoch_adcs_mat  = np.array(first_adcs_all[:, padded_col])    # (N_RUNS, N_common)
Nticks_wf = wfs.shape[1]

# Valid stochastic hit: tick within waveform bounds
valid_hit = stoch_ticks_mat < (Nticks_wf - 3)               # (N_RUNS, N_common)

# ── 3. Build window tick indices (N_RUNS, N_common, 2*TICK_TOL+1) ──────────────
W = 2 * TICK_TOL + 1
delta = np.arange(-TICK_TOL, TICK_TOL + 1)                  # (W,)
# window_ticks[r, p, w] = stoch_ticks_mat[r, p] + delta[w], clipped to [0, Nticks-1]
window_ticks = np.clip(
    stoch_ticks_mat[:, :, None] + delta[None, None, :],
    0, Nticks - 2
).astype(int)   # (N_RUNS, N_common, W)

# ── 4. Gather t_prob and a_exp at window ticks ─────────────────────────────────
# t_prob_mat[col_indices[p], window_ticks[r, p, w]]
# We broadcast using advanced indexing over the pixel dimension
col_idx_bcast = col_indices[None, :, None]                  # (1, N_common, 1)
window_probs  = t_prob_mat[col_idx_bcast, window_ticks]     # (N_RUNS, N_common, W)
window_adcs   = a_exp_mat[col_idx_bcast, window_ticks]      # (N_RUNS, N_common, W)

# ── 5. Weighted-average expected ADC over window ────────────────────────────────
w_sum = window_probs.sum(axis=-1)                            # (N_RUNS, N_common)
safe_w_sum = np.where(w_sum > 1e-12, w_sum, 1.0)
prob_adc_matched = (window_probs * window_adcs).sum(axis=-1) / safe_w_sum  # (N_RUNS, N_common)

# ── 6. Flatten and filter to valid hits with non-zero weight ───────────────────
flat_valid  = valid_hit.ravel() & (w_sum.ravel() > 1e-12)
stoch_adc_flat = stoch_adcs_mat.ravel()[flat_valid]
prob_adc_flat  = prob_adc_matched.ravel()[flat_valid]

# Convert ADC → charge [ke-]
stoch_q = np.array(adc2charge(stoch_adc_flat, current_params))
prob_q  = np.array(adc2charge(prob_adc_flat,  current_params))
dq      = prob_q - stoch_q

print(f"Matched hit pairs : {len(dq):,}")
print(f"Mean ΔQ (prob−stoch) [ke-]: {dq.mean():.4f}")
print(f"Std  ΔQ              [ke-]: {dq.std():.4f}")

# ── 7. Plots ────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(13, 4))

lim = max(np.abs(stoch_q).max(), np.abs(prob_q).max()) * 1.05
# Subsample for the scatter plot to avoid over-plotting
rng = np.random.default_rng(0)
plot_idx = rng.choice(len(dq), size=min(50_000, len(dq)), replace=False)
axes[0].scatter(stoch_q[plot_idx], prob_q[plot_idx], s=2, alpha=0.10)
axes[0].plot([0, lim], [0, lim], 'r--', lw=1, label='y=x')
axes[0].set_xlabel('Stochastic Q [ke-]')
axes[0].set_ylabel('Probabilistic expected Q [ke-]')
axes[0].set_title('Charge: probabilistic vs stochastic (matched hits)')
axes[0].legend(fontsize=8)

axes[1].hist(dq, bins=120, edgecolor='k', linewidth=0.3)
axes[1].axvline(dq.mean(), color='r', linestyle='--',
                label=f'mean={dq.mean():.4f} ke-')
axes[1].set_xlabel('ΔQ = prob − stoch [ke-]')
axes[1].set_ylabel('Count')
axes[1].set_title('Charge residuals (matched hits)')
axes[1].legend()

plt.tight_layout()
plt.show()
```

## Formal Distribution Comparison: Stochastic vs. Probabilistic (per pixel)

For every pixel we compare the probabilistic predicted tick distribution $P_{\text{prob}}(t)$ against the empirical stochastic histogram $\hat{P}_{\text{stoch}}(t)$ using three complementary metrics:

| Metric | Formula | What it measures |
|---|---|---|
| **TVD** (Total Variation) | $\tfrac{1}{2}\sum_t|P_{\text{prob}}-\hat{P}_{\text{stoch}}|$ | Overall L1 mismatch, $\in[0,1]$ |
| **KL divergence** | $\sum_t \hat{P}_{\text{stoch}}\log\frac{\hat{P}_{\text{stoch}}}{P_{\text{prob}}}$ | Log-prob loss of the probabilistic model on stochastic data |
| **Wasserstein-1** | $\sum_t|\text{CDF}_{\text{prob}}-\text{CDF}_{\text{stoch}}|$ | Earth-mover (respects tick ordering) |

All three are computed fully vectorised over the pixel dimension using the empirical tick histogram matrix built directly from `stoch_ticks_mat`.

```python
# --- Cell G: Formal distribution comparison metrics (TVD, KL, Wasserstein-1) ---
# NOTE: Reset to reference parameters so Cell G is independent of any tuning
# that may have been applied in the sweep cell above. The comparison is run on
# the raw probabilistic model to expose the true approximation error.

# ── 0. Ensure we use the untuned reference parameters ─────────────────────────
_params_g = ref_params.replace(**initial_values)   # clean slate, no excess_offset / sigma tweaks
if _params_g != current_params:
    print("[Cell G] Rebuilding probabilistic distribution with reference parameters ...")
    from larndsim.sim_jax import simulate_probabilistic
    _, _, _, _ticks_prob_g, _ = simulate_probabilistic(_params_g, wfs, unique_pixels)
    _t_prob_mat_g_full = np.zeros((wfs.shape[0], wfs.shape[1] - 1), dtype=np.float64)
    for _pi, _tp in enumerate(unique_pixels):
        if int(_tp) < 0:
            continue
        _t_prob_mat_g_full[_pi] = np.array(jnp.exp(_ticks_prob_g[_pi, 0, :]))
    # Restrict to valid (non-padding) pixels, same mask as Cell D
    _t_prob_mat_g = _t_prob_mat_g_full[valid_indices]
    print(f"[Cell G] Done. excess_offset={_params_g.excess_offset}, "
          f"smooth_sigma_scale={_params_g.smooth_sigma_scale}")
else:
    _t_prob_mat_g = t_prob_mat   # already using reference params
    print(f"[Cell G] current_params already matches reference — reusing existing t_prob_mat.")
    print(f"         excess_offset={_params_g.excess_offset}, smooth_sigma_scale={_params_g.smooth_sigma_scale}")

# Prereqs: common_pixels, col_indices, padded_col, t_prob_mat (from Cell D)
#          stoch_ticks_mat, valid_hit  (from Cell D)

# ── 1. Build stochastic empirical histogram (N_common, Nticks) ─────────────────
Nticks_hist = _t_prob_mat_g.shape[1]
N_common = len(common_pixels)

stoch_hist_raw = np.zeros((N_common, Nticks_hist), dtype=np.float64)
for p_idx in range(N_common):
    valid_runs = valid_hit[:, p_idx]                         # (N_RUNS,) bool
    ticks_p = stoch_ticks_mat[valid_runs, p_idx].astype(np.int32)
    stoch_hist_raw[p_idx] = np.bincount(
        np.clip(ticks_p, 0, Nticks_hist - 1), minlength=Nticks_hist
    )

# Normalize: conditional distribution over ticks given a hit occurred
hit_counts = valid_hit.sum(axis=0).astype(np.float64)       # (N_common,)
safe_counts = np.where(hit_counts > 0, hit_counts, 1.0)
stoch_hist = stoch_hist_raw / safe_counts[:, None]           # (N_common, Nticks)

# Probabilistic distribution for the same pixels
t_prob_common_g = _t_prob_mat_g[col_indices]                 # (N_common, Nticks)

# Normalize both to sum to 1 for conditional comparison
t_prob_norm = t_prob_common_g / np.maximum(t_prob_common_g.sum(axis=1, keepdims=True), 1e-30)
stoch_norm  = stoch_hist       / np.maximum(stoch_hist.sum(axis=1, keepdims=True),     1e-30)

# ── 2. TVD (vectorised) ───────────────────────────────────────────────────────
tvd = 0.5 * np.abs(t_prob_norm - stoch_norm).sum(axis=1)   # (N_common,)

# ── 3. KL divergence KL(stoch || prob) (vectorised) ──────────────────────────
eps_kl = 1e-30
log_ratio = np.log(np.maximum(stoch_norm, eps_kl) / np.maximum(t_prob_norm, eps_kl))
kl = (stoch_norm * log_ratio).sum(axis=1)                   # (N_common,)
kl = np.where(hit_counts >= 10, kl, np.nan)                 # unreliable for tiny samples

# ── 4. Wasserstein-1 via CDF trick (vectorised) ───────────────────────────────
# W1 = Σ_t |CDF_prob(t) - CDF_stoch(t)|   (discrete, equal-spaced bins → units: ticks)
cdf_prob  = np.cumsum(t_prob_norm, axis=1)
cdf_stoch = np.cumsum(stoch_norm,  axis=1)
wass1 = np.abs(cdf_prob - cdf_stoch).sum(axis=1)            # (N_common,) [ticks]
wass1 = np.where(hit_counts >= 10, wass1, np.nan)

# ── 5. Mean tick shift (first-moment bias) ────────────────────────────────────
tick_range = np.arange(Nticks_hist, dtype=np.float64)
mean_tick_prob  = (t_prob_norm * tick_range[None, :]).sum(axis=1)
mean_tick_stoch = (stoch_norm  * tick_range[None, :]).sum(axis=1)
tick_shift = mean_tick_prob - mean_tick_stoch               # (N_common,) [ticks]
tick_shift = np.where(hit_counts >= 10, tick_shift, np.nan)

# ── 6. Summary ────────────────────────────────────────────────────────────────
valid_g = np.isfinite(kl)
print(f"Pixels with ≥10 stochastic hits : {valid_g.sum()} / {N_common}")
print(f"TVD          mean={np.nanmean(tvd):.4f}  median={np.nanmedian(tvd):.4f}  max={np.nanmax(tvd):.4f}")
print(f"KL(stoch||p) mean={np.nanmean(kl):.4f}  median={np.nanmedian(kl):.4f}  max={np.nanmax(kl):.4f}")
print(f"Wasserstein1 mean={np.nanmean(wass1):.2f} ticks  median={np.nanmedian(wass1):.2f}  max={np.nanmax(wass1):.2f}")
print(f"Mean tick shift mean={np.nanmean(tick_shift):.3f} ticks  (+ = prob predicts later)")

# ── 7. Plots ──────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(2, 2, figsize=(13, 8))

axes[0, 0].hist(tvd[valid_g], bins=40, edgecolor='k', linewidth=0.4, color='steelblue')
axes[0, 0].axvline(np.nanmedian(tvd), color='r', linestyle='--',
                   label=f'median={np.nanmedian(tvd):.3f}')
axes[0, 0].set_xlabel('TVD'); axes[0, 0].set_ylabel('Pixels')
axes[0, 0].set_title('Total Variation Distance'); axes[0, 0].legend()

axes[0, 1].hist(kl[valid_g], bins=40, edgecolor='k', linewidth=0.4, color='darkorange')
axes[0, 1].axvline(np.nanmedian(kl), color='r', linestyle='--',
                   label=f'median={np.nanmedian(kl):.3f}')
axes[0, 1].set_xlabel('KL(stoch || prob) [nats]'); axes[0, 1].set_ylabel('Pixels')
axes[0, 1].set_title('KL Divergence'); axes[0, 1].legend()

axes[1, 0].hist(wass1[valid_g], bins=40, edgecolor='k', linewidth=0.4, color='mediumseagreen')
axes[1, 0].axvline(np.nanmedian(wass1), color='r', linestyle='--',
                   label=f'median={np.nanmedian(wass1):.2f} ticks')
axes[1, 0].set_xlabel('Wasserstein-1 [ticks]'); axes[1, 0].set_ylabel('Pixels')
axes[1, 0].set_title('Wasserstein-1 Distance'); axes[1, 0].legend()

axes[1, 1].hist(tick_shift[valid_g], bins=40, edgecolor='k', linewidth=0.4, color='mediumpurple')
axes[1, 1].axvline(np.nanmean(tick_shift), color='r', linestyle='--',
                   label=f'mean={np.nanmean(tick_shift):.3f} ticks')
axes[1, 1].axvline(0, color='k', linestyle=':', lw=1)
axes[1, 1].set_xlabel('Mean tick shift (prob − stoch) [ticks]'); axes[1, 1].set_ylabel('Pixels')
axes[1, 1].set_title('First-Moment Bias (timing)'); axes[1, 1].legend()

plt.suptitle('Formal distribution comparison: probabilistic vs. stochastic (per pixel, first hit)\n'
             f'(reference params: excess_offset={_params_g.excess_offset:.0f}, '
             f'smooth_sigma_scale={_params_g.smooth_sigma_scale:.2f})',
             fontsize=12)
plt.tight_layout()
plt.show()

# ── 8. Worst pixels ───────────────────────────────────────────────────────────
print("\n── 5 worst pixels by TVD ──")
for i in np.argsort(tvd)[::-1][:5]:
    p = common_pixels[i]
    print(f"  pixel={p:8d}  TVD={tvd[i]:.4f}  KL={kl[i]:.4f}  "
          f"W1={wass1[i]:.2f}  tick_shift={tick_shift[i]:.2f}  "
          f"emp_rate={emp_rate[i]:.3f}  lambda={lambda_prob[i]:.3f}")

print("\n── 5 worst pixels by Wasserstein-1 ──")
for i in np.argsort(np.where(np.isfinite(wass1), wass1, -1))[::-1][:5]:
    p = common_pixels[i]
    print(f"  pixel={p:8d}  TVD={tvd[i]:.4f}  KL={kl[i]:.4f}  "
          f"W1={wass1[i]:.2f}  tick_shift={tick_shift[i]:.2f}  "
          f"emp_rate={emp_rate[i]:.3f}  lambda={lambda_prob[i]:.3f}")
```

## Diagnostic I: Survival-Factor Ratio

The beam-search proxy for P(first crossing at t) does **not** include the survival factor
$\prod_{s<t}(1-\Phi(z_s))$ that the exact first-crossing probability requires.

We can measure the overall magnitude of this missing factor for each pixel by comparing
$\lambda_{\text{prob}} = \sum_t P(t)$ against $\Phi\!\left(\frac{q_{\text{peak}}-\Theta}{\sigma_r}\right)$,
the probability that the threshold is *ever* exceeded (an upper bound on the true hit probability).

$$R = \frac{\lambda_{\text{prob}}}{\Phi(z_{\text{peak}})}$$

- $R\approx 1$: approximation is tight (monotone, steep signals).
- $R\ll 1$: survival-factor under-estimation is severe (marginal / bipolar waveforms).
- $R\gg 1$: the proxy over-counts (overlapping windows, seen for $\lambda>1$ pixels).

We also compare the ratio against the empirical hit rate to confirm it predicts the bias direction.

```python
# --- Cell I: Survival-factor ratio diagnostic ---
# Prereqs: common_pixels, lambda_prob, emp_rate, all_tick_probs, wfs,
#          prob_prediction, current_params  (from earlier cells)
# Also uses _params_g and _t_prob_mat_g from Cell G (reference-param t_prob).

import scipy.stats as ss

sigma_r = float(current_params.RESET_NOISE_CHARGE)
threshold = float(current_params.DISCRIMINATION_THRESHOLD)

# ── Per-pixel: lambda_prob (using reference params from Cell G) ──
# If Cell G rebuilt _t_prob_mat_g with reference params, use that;
# otherwise fall back to all_tick_probs which may be tuned.
_lambda_ref = np.array([
    float(_t_prob_mat_g[col_indices[i]].sum()) for i in range(len(common_pixels))
])

# ── Per-pixel: P(threshold ever exceeded) = Φ(z_peak) ──
# z_peak = (max(q_cum) − threshold) / sigma_r
q_cum_peak = np.zeros(len(common_pixels))
for i, p in enumerate(common_pixels):
    pidx = int(jnp.where(prob_prediction['unique_pixels'] == p)[0][0])
    wf_p = np.array(wfs[pidx])
    q_cum_p = np.cumsum(wf_p * float(current_params.t_sampling))
    q_cum_peak[i] = q_cum_p.max()

z_peak = (q_cum_peak - threshold) / sigma_r
prob_ever_cross = ss.norm.cdf(z_peak)              # Φ(z_peak)

# ── Survival-factor ratio ──
survival_ratio = np.where(prob_ever_cross > 1e-6, _lambda_ref / prob_ever_cross, np.nan)

print("Survival-factor ratio R = λ_prob / Φ(z_peak)  (should be ≈1 for well-behaved pixels)")
print(f"  R mean  = {np.nanmean(survival_ratio):.4f}")
print(f"  R median= {np.nanmedian(survival_ratio):.4f}")
print(f"  R min   = {np.nanmin(survival_ratio):.4f}")
print(f"  R max   = {np.nanmax(survival_ratio):.4f}")

# ── Plots ──
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

# Left: R histogram
valid_r = np.isfinite(survival_ratio)
axes[0].hist(survival_ratio[valid_r], bins=40, edgecolor='k', linewidth=0.4, color='mediumslateblue')
axes[0].axvline(1.0, color='r', linestyle='--', label='R=1 (ideal)')
axes[0].axvline(np.nanmedian(survival_ratio), color='orange', linestyle='-.',
                label=f'median={np.nanmedian(survival_ratio):.3f}')
axes[0].set_xlabel(r'$R = \lambda_{\mathrm{prob}} \,/\, \Phi(z_{\mathrm{peak}})$')
axes[0].set_ylabel('Pixels')
axes[0].set_title('Survival-factor ratio distribution')
axes[0].legend(fontsize=8)

# Middle: R vs. bias (lambda - emp)
bias_g = _lambda_ref - emp_rate
sc = axes[1].scatter(survival_ratio[valid_r], bias_g[valid_r],
                     c=emp_rate[valid_r], cmap='viridis', s=20, alpha=0.7)
plt.colorbar(sc, ax=axes[1], label='Empirical hit rate')
axes[1].axhline(0, color='k', linestyle=':', lw=1)
axes[1].axvline(1, color='r', linestyle='--', lw=1)
axes[1].set_xlabel(r'$R = \lambda_{\mathrm{prob}} \,/\, \Phi(z_{\mathrm{peak}})$')
axes[1].set_ylabel(r'Bias $= \lambda_{\mathrm{prob}} - \mathrm{emp\_rate}$')
axes[1].set_title('Bias vs. survival-factor ratio')

# Right: R vs. z_peak (steepness proxy)
axes[2].scatter(z_peak[valid_r], survival_ratio[valid_r],
                c=bias_g[valid_r], cmap='RdBu_r', s=20, alpha=0.7,
                vmin=-0.15, vmax=0.15)
plt.colorbar(plt.cm.ScalarMappable(cmap='RdBu_r',
                norm=plt.Normalize(-0.15, 0.15)), ax=axes[2], label='Bias')
axes[2].axhline(1.0, color='r', linestyle='--', lw=1, label='R=1')
axes[2].set_xlabel(r'$z_{\mathrm{peak}} = (q_{\mathrm{cum,max}} - \Theta) / \sigma_r$')
axes[2].set_ylabel(r'$R$')
axes[2].set_title('Survival-factor ratio vs. signal steepness')
axes[2].legend(fontsize=8)

plt.suptitle('Survival-factor ratio: how well the beam-search proxy covers \u03a6(z_peak)',
             fontsize=11)
plt.tight_layout()
plt.show()

# Print worst pixels by survival-factor deviation
print("\n── Pixels with most severe under-estimation (R < 0.7) ──")
for i in np.argsort(survival_ratio):
    if not valid_r[i] or survival_ratio[i] >= 0.7:
        continue
    p = common_pixels[i]
    print(f"  pixel={p:8d}  R={survival_ratio[i]:.3f}  λ={_lambda_ref[i]:.3f}  "
          f"emp={emp_rate[i]:.3f}  bias={bias_g[i]:+.3f}  z_peak={z_peak[i]:.2f}")

print("\n── Pixels with over-estimation (R > 1.2) ──")
for i in np.argsort(survival_ratio)[::-1]:
    if not valid_r[i] or survival_ratio[i] <= 1.2:
        break
    p = common_pixels[i]
    print(f"  pixel={p:8d}  R={survival_ratio[i]:.3f}  λ={_lambda_ref[i]:.3f}  "
          f"emp={emp_rate[i]:.3f}  bias={bias_g[i]:+.3f}  z_peak={z_peak[i]:.2f}")
```

## Diagnostic J: Bias vs. Slope of Q_cum at Threshold

The beam-search approximation error is largest when the cumulative charge rises *slowly* through the
discrimination threshold — because a slow rise means the crossing probability is spread over many
ticks, amplifying the survival-factor error.

We measure the "threshold slope" as:
$$s = \frac{\Delta q_{\text{cum}}}{\Delta t}\Bigg|_{q_{\text{cum}}\approx\Theta}$$
by taking the median derivative of $q_{\text{cum}}$ in the window where it is within $\pm 2\sigma_r$
of the threshold. Steep slopes (large $s$) correspond to a sharply rising signal that triggers reliably;
shallow slopes (small $s$) correspond to marginal pixels.

Expected behaviour: **bias should be more negative (under-estimation) for small $s$.**

```python
# --- Cell J: Bias vs. slope of Q_cum at threshold ---
# Prereqs: common_pixels, lambda_prob, emp_rate, wfs, prob_prediction,
#          current_params, bias_g, _lambda_ref  (from Cells A and I)

from scipy.ndimage import uniform_filter1d

sigma_r   = float(current_params.RESET_NOISE_CHARGE)
threshold = float(current_params.DISCRIMINATION_THRESHOLD)
t_samp    = float(current_params.t_sampling)    # seconds per tick

# ── Compute threshold slope for each pixel ──
slope_at_threshold = np.full(len(common_pixels), np.nan)  # e-/tick
q_peak_arr        = np.full(len(common_pixels), np.nan)

for i, p in enumerate(common_pixels):
    pidx = int(jnp.where(prob_prediction['unique_pixels'] == p)[0][0])
    wf_p = np.array(wfs[pidx])
    q_cum_p = np.cumsum(wf_p * t_samp)

    # Find ticks where q_cum is within [θ − 2σ_r, θ + 2σ_r]
    near_threshold = np.where(
        (q_cum_p >= threshold - 2 * sigma_r) & (q_cum_p <= threshold + 2 * sigma_r)
    )[0]
    q_peak_arr[i] = q_cum_p.max()

    if len(near_threshold) < 2:
        # Threshold not reached or crossed too abruptly: use derivative at
        # the tick closest to the threshold.
        t_closest = int(np.argmin(np.abs(q_cum_p - threshold)))
        # Finite difference over a ±3-tick window
        lo, hi = max(0, t_closest - 3), min(len(q_cum_p) - 1, t_closest + 3)
        if hi > lo:
            slope_at_threshold[i] = (q_cum_p[hi] - q_cum_p[lo]) / (hi - lo)
        continue

    # Use the gradient within the near-threshold window
    # Smooth Q_cum slightly before differentiating to reduce noise
    q_smooth = uniform_filter1d(q_cum_p, size=3)
    dq = np.diff(q_smooth)          # [e-/tick]
    slope_at_threshold[i] = np.median(dq[near_threshold[:-1]])

# Convert to ke-/tick for display
slope_ke_tick = slope_at_threshold * 1e-3

print("Threshold slope s [ke-/tick]:")
print(f"  mean   = {np.nanmean(slope_ke_tick):.4f}")
print(f"  median = {np.nanmedian(slope_ke_tick):.4f}")
print(f"  min    = {np.nanmin(slope_ke_tick):.4f}")
print(f"  max    = {np.nanmax(slope_ke_tick):.4f}")

# ── Plots ──
valid_s = np.isfinite(slope_ke_tick) & np.isfinite(bias_g)

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

# Left: slope histogram
axes[0].hist(slope_ke_tick[valid_s], bins=40, edgecolor='k', linewidth=0.4, color='darkcyan')
axes[0].set_xlabel('Threshold slope $s$ [ke$^-$/tick]')
axes[0].set_ylabel('Pixels')
axes[0].set_title('Distribution of threshold slope')

# Middle: bias vs. slope  (main diagnostic plot)
sc2 = axes[1].scatter(
    slope_ke_tick[valid_s], bias_g[valid_s],
    c=emp_rate[valid_s], cmap='viridis', s=25, alpha=0.8
)
plt.colorbar(sc2, ax=axes[1], label='Empirical hit rate')
axes[1].axhline(0, color='k', linestyle=':', lw=1)
axes[1].set_xlabel('Threshold slope $s$ [ke$^-$/tick]')
axes[1].set_ylabel(r'Bias $= \lambda_{\mathrm{prob}} - \mathrm{emp\_rate}$')
axes[1].set_title('Bias vs. threshold slope\n(neg. bias → under-estimation at slow rise)')

# Annotate the worst-bias pixel
worst_idx = int(np.nanargmin(bias_g))
if valid_s[worst_idx]:
    axes[1].annotate(
        f'pix {common_pixels[worst_idx]}',
        xy=(slope_ke_tick[worst_idx], bias_g[worst_idx]),
        xytext=(slope_ke_tick[worst_idx] + 0.02, bias_g[worst_idx] - 0.01),
        fontsize=7, arrowprops=dict(arrowstyle='->', lw=0.8)
    )

# Right: survival-factor ratio vs. slope (shows the coupling)
if 'survival_ratio' in dir():
    valid_sr = valid_s & valid_r
    sc3 = axes[2].scatter(
        slope_ke_tick[valid_sr], survival_ratio[valid_sr],
        c=emp_rate[valid_sr], cmap='viridis', s=25, alpha=0.8
    )
    plt.colorbar(sc3, ax=axes[2], label='Empirical hit rate')
    axes[2].axhline(1.0, color='r', linestyle='--', lw=1, label='R=1 (ideal)')
    axes[2].set_xlabel('Threshold slope $s$ [ke$^-$/tick]')
    axes[2].set_ylabel(r'$R = \lambda_{\mathrm{prob}} / \Phi(z_{\mathrm{peak}})$')
    axes[2].set_title('Survival-factor ratio vs. threshold slope')
    axes[2].legend(fontsize=8)
else:
    axes[2].text(0.5, 0.5, 'Run Cell I first\nto get survival_ratio',
                 ha='center', va='center', transform=axes[2].transAxes)
    axes[2].set_visible(True)

plt.suptitle('Threshold slope: probing where the beam-search approximation breaks down',
             fontsize=11)
plt.tight_layout()
plt.show()

# ── Tabulate per-pixel summary for pixels near the boundary ──
print("\n── All pixels with |bias| > 0.03, sorted by slope ──")
header = f"{'pixel':>10s}  {'slope [ke-/t]':>14s}  {'lambda':>7s}  {'emp':>7s}  {'bias':>8s}  {'R':>6s}"
print(header)
for i in np.argsort(slope_ke_tick):
    if not valid_s[i]:
        continue
    if abs(bias_g[i]) < 0.03:
        continue
    R_str = f"{survival_ratio[i]:.3f}" if ('survival_ratio' in dir() and valid_r[i]) else '  N/A'
    print(f"  {common_pixels[i]:>8d}  {slope_ke_tick[i]:>14.4f}  "
          f"{_lambda_ref[i]:>7.4f}  {emp_rate[i]:>7.4f}  "
          f"{bias_g[i]:>+8.4f}  {R_str:>6s}")
```
